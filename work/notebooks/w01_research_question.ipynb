## 2. The question: decision, action, cost of a wrong call

### Research Question

Which content pages should be reviewed first for refresh based on their
    search performance and engagement signals?

### Decision

The decision is to identify and prioritize content pages that should be reviewed before others.

### Action

A content 
    reviewer or SEO team will review the recommended pages and decide whether they should be refreshed, expanded, monitored, or improved.

### Unit of Analysis

Each row in the dataset 
    represents one content page.

### Output

The expected output is a ranked list of content pages with refresh scores and reason codes explaining why each page was recommended.

### 
    Cost of a Wrong Recommendation

If a page is incorrectly recommended (false positive), reviewers spend time reviewing a page that may not need attention.

If an important declining page 
    is missed (false negative), its search performance may continue to decrease, resulting in lost traffic and missed opportunities.

### Why Data and Machine Learning Help

The dataset contains
    multiple search and engagement signals that are difficult to evaluate manually across thousands of pages. Data analysis and machine learning can combine these signals to rank pages for review,
    helping reviewers make faster and more consistent decisions.

## 3. Quick look at the data (2-3 real numbers)
